```python
import numpy as np
#just importin numpy and then using it through this 'np' alias that we gave
```

```python
arr = np.array([2, 9, 9, 7, 9, 2, 4, 5, 8])
# creating a numpy array from a Python list

print(arr)
```

```python
rarr = arr[::-1]
print(rarr)
# slicing trick to reverse the array
# [start:end:step] → step = -1 means go backwards
```

```python
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[1,2,3],[4,5,6]])
# creating two 2D arrays (basically matrices)

print(arr1 == arr2)
# element-wise comparison
# compares each position and returns True/False
# not checking if whole arrays are equal, but each element
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1,1,1,2,3,4,2,4,3,3])
# two arrays with repeated values

mfreq_x = np.argmax(np.bincount(x))
# bincount → counts frequency of each number
# argmax → index of the highest frequency
# so this gives most frequent number in x

mfreq_y = np.argmax(np.bincount(y))
# same idea for y

print(mfreq_x, mfreq_y)
# prints most frequent elements
# output: 1 1
```

```python
anotherArray = np.matrix('[4,1,9;12,3,1;4,5,6]')
# creating a matrix using numpy's matrix class
# semicolons separate rows

print(anotherArray.sum(), '\n', anotherArray.sum(axis=1), '\n', anotherArray.sum(axis=0))
# anotherArray.sum() → sum of all elements
# axis=1 → row-wise sum
# axis=0 → column-wise sum
```
