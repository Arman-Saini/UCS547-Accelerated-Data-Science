```python
import numpy as np
import random
import time
# numpy for fast array ops
# random for generating numbers
# time to measure how long things take
```

```python
N = 1_000_000
# size of data (1 million)
```

```python
# ---------- create Python list ----------
l1 = []
for i in range(N):
    l1.append(random.randint(1,10000))
# normal Python list filled with random integers
```

```python
# ---------- create NumPy array ----------
l2 = np.zeros(N)
for i in range(N):
    l2[i] = random.randint(1,10000)
# numpy array filled using loop (not the fastest way,
# but keeps things fair since both used loops)
```

```python
# ---------- timing Python list squaring ----------
start = time.time()
# start timer

l3 = []
for i in range(N):
    l3.append(l1[i] * l1[i])
# square each element using Python loop

end = time.time()
print("Python list loop time:", end - start, "seconds")
# prints time taken
```

```python
# ---------- timing NumPy vectorized squaring ----------
start = time.time()
# start timer

c = l2 * l2
# numpy vectorized operation (very fast)

end = time.time()
print("NumPy vectorized time:", end - start, "seconds")
# prints time taken
```
